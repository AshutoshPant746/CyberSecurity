```python
from waybackpy import WaybackMachineCDXServerAPI

def extract_archived_urls(url, start_year, end_year, limit=10):
    """Retrieve archived URLs within a date range."""
    cdx = WaybackMachineCDXServerAPI(url, start_timestamp=str(start_year), end_timestamp=str(end_year))
    snapshots = list(cdx.snapshots())[:limit]

    print(f"🔍 Extracting archived URLs for {url} from {start_year} to {end_year}\n")
    for idx, snapshot in enumerate(snapshots, 1):
        print(f"[{idx}] {snapshot.archive_url}")

# Example Run:
extract_archived_urls("upes.ac.in", 2018, 2020, 10)
```

```python
from googlesearch import search

def google_dork_deleted_pages(domain):
    query = f"site:{domain} -site:web.archive.org"

    print(f"Searching for deleted pages on: {domain}\n")

    try:
        results = search(query, num_results=5)  # Get top 5 results
        if results:
            for url in results:
                print(url)
        else:
            print("No deleted pages found.")
    except Exception as e:
        print(f"Error: {e}")

# Example usage
google_dork_deleted_pages("upes.ac.in")
```

```python
import time
from googlesearch import search

def find_pdfs_google(website, start_year, end_year, limit=5, delay=5):
    """Find PDF documents indexed by Google on a university website."""
    query = f"site:{website} filetype:pdf after:{start_year} before:{end_year}"

    print(f"\nSearching for PDFs on {website} from {start_year} to {end_year}\n")

    try:
        results = search(query)  # No extra arguments
        results_list = list(results)[:limit]  # Manually limiting results

        for idx, url in enumerate(results_list, 1):
            print(f"[{idx}] 📄 {url}")
            time.sleep(delay)  # Delay to avoid rate limiting

    except Exception as e:
        print(f"Error: {e}")

# Example Run:
find_pdfs_google("stanford.edu", 2015, 2022, 5)
```

```python
from waybackpy import WaybackMachineCDXServerAPI

def track_website_changes(url, years):
    """Fetch archived snapshots for specific years to track website changes."""
    print(f"Tracking changes for {url} over different years\n")

    for year in years:
        cdx = WaybackMachineCDXServerAPI(url, start_timestamp=f"{year}", end_timestamp=f"{year}")
        snapshots = list(cdx.snapshots())

        if snapshots:
            print(f"{year}: {snapshots[0].archive_url}")
        else:
            print(f"No snapshot found for {year}.")

# Example Run:
track_website_changes("nasa.gov", [2010, 2013, 2016, 2019, 2022])
```

```python
from googlesearch import search
import time

def find_admin_login_pages(website, keywords, limit=5, delay=5):
    print(f"Archived URLs Containing 'admin' or 'login' for {website}\n")

    # Construct the search query
    query = f"site:{website} ({' OR '.join(keywords)})"

    try:
        # Use search function and manually limit the results
        search_results = list(search(query))[:limit]

        # Print the first `limit` matching pages
        for idx, result in enumerate(search_results, 1):
            print(f"{idx}. {result}")
            time.sleep(delay)  # Delay to avoid rate limiting

    except Exception as e:
        print(f"An error occurred: {e}")

# Define the website and keywords
website = "upes.ac.in"
keywords = ["admin", "login"]

# Find and display the matching pages
find_admin_login_pages(website, keywords)
```
